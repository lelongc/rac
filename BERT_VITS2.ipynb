🪄 🟢 Ô CODE SỐ 3: PHÉP THUẬT DỌN NHÀ (Dành riêng cho sếp)
Trong video không có ô này. Ô này đệ tử viết riêng để tự động lấy kurisu_final.list và đống âm thanh của sếp, sửa lại đúng chuẩn format của Style-Bert-VITS2 mà sếp không cần đụng tay vào.

```python
import os
import shutil

# Đường dẫn gốc của sếp
old_list = "/content/drive/MyDrive/ggcl/kurisu/kurisu_final.list"
old_wav_dir = "/content/drive/MyDrive/ggcl/kurisu/dataset_wav"

# Nơi ở mới của Style-Bert-VITS2
model_name = "kurisu"
dataset_root = "/content/drive/MyDrive/Style-Bert-VITS2/Data"
model_dir = os.path.join(dataset_root, model_name)
raw_dir = os.path.join(model_dir, "raw")
esd_path = os.path.join(model_dir, "esd.list")

os.makedirs(raw_dir, exist_ok=True)
os.makedirs("/content/drive/MyDrive/Style-Bert-VITS2/model_assets", exist_ok=True)

print("1. Đang định dạng lại file list và LỌC SẠCH DẤU CÂU cho sếp...")
with open(old_list, 'r', encoding='utf-8') as f_old, open(esd_path, 'w', encoding='utf-8') as f_new:
    for line in f_old:
        line = line.strip()
        if not line: continue
        parts = line.split('|')
        if len(parts) >= 4:
            wav_name = os.path.basename(parts[0])
            speaker = parts[1]
            text = parts[3]

            # --- BỘ LỌC TỰ ĐỘNG CHUYỂN DẤU TIẾNG ANH SANG TIẾNG NHẬT ---
            text = text.replace(',', '、').replace('?', '？').replace('!', '！').replace('.', '。')

            f_new.write(f"{wav_name}|{speaker}|JP|{text}\n")

print("2. Đang tự động Copy toàn bộ file âm thanh... (Chờ khoảng 1-2 phút nhé sếp)...")
for file in os.listdir(old_wav_dir):
    if file.endswith(".wav"):
        shutil.copy2(os.path.join(old_wav_dir, file), os.path.join(raw_dir, file))

print("✅ ĐÃ LỌC SẠCH DẤU CÂU VÀ CHUẨN BỊ DATA XONG! SẾP CHẠY Ô TIẾP THEO!")
```
