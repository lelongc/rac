d. Giả sử bộ dữ liệu là toàn bộ kết quả điểm thi SAT của học sinh một trường phố thông. Sử dụng mẫu là điểm thi của 36 học sinh đầu tiên trong bộ dữ liệu, bạn hãy xây dựng khoảng tin cậy **99%** cho điểm SAT trung bình của trường phổ thông đấy.

e. Giả sử bộ dữ liệu là toàn bộ kết quả điểm thi SAT của học sinh một trường phố thông. Sử dụng mẫu là điểm thi của 36 học sinh đầu tiên trong bộ dữ liệu, bạn hãy xây dựng khoảng tin cậy **95%** cho điểm SAT trung bình của trường phổ thông đấy. Cho biết khoảng ước lượng bạn xây dựng có đúng không, dựa trên giá trị đã tính ở câu trên.

```python
import pandas as pd
import numpy as np
import scipy.stats as st
# Đọc dữ liệu
data = pd.read_csv('SAT-lard.csv', sep=',')  # hoặc ',' tùy định dạng
scores = data['SAT Score']

n = len(scores)
xbar = np.mean(scores)
s = np.std(scores, ddof=1)

print(f"Số quan sát n = {n}")
print(f"Trung bình mẫu = {xbar:.2f}")
print(f"Độ lệch chuẩn mẫu = {s:.2f}")
```

```python
conf = 0.99
z = st.norm.ppf(1 - (1 - conf)/2)
moe = z * s / np.sqrt(n)
(float(xbar - moe), float(xbar + moe))
```

```python
conf = 0.95
z = st.norm.ppf(1 - (1 - conf)/2)
moe = z * s / np.sqrt(n)
(float(xbar - moe), float(xbar + moe))
```

```python
mu = np.mean(scores)
float(mu)
```

```python
sample = scores[:36]
n = len(sample)
xbar = np.mean(sample)
s = np.std(sample, ddof=1)
conf = 0.99
t = st.t.ppf(1 - (1 - conf)/2, df=n-1)
moe = t * s / np.sqrt(n)
(float(xbar - moe), float(xbar + moe))
```

```python
conf = 0.95
t = st.t.ppf(1 - (1 - conf)/2, df=n-1)
moe = t * s / np.sqrt(n)
ci_95 = (float(xbar - moe), float(xbar + moe))
ci_95
```
