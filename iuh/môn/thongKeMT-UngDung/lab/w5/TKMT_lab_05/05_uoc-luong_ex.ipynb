### Bài 1: Tạo quần thể mô phỏng

Bạn hãy tạo một mẫu gồm 10000 phần tử mô phỏng chiều cao của nam thanh niên có giá trị từ 120cm - 200cm. Bạn hãy lưu kết quả vào biến **POP**.

Tính:
a. Chiều cao trung bình (kỳ vọng) của quần thể và độ lệch chuẩn về chiều cao của quần thể.

b. Tính tỷ lệ người cao trong quần thể, biết rằng thanh niên có chiều cao từ 180 trở lên được gọi là cao.

c. Vẽ histogram về chiều cao của quần thể. Theo bạn quần thể có phân phối chuẩn hay không?

d. Thử vẽ histogram và đồ thị hàm mật độ của phân phối chuẩn sử dụng tham số loc và scale bằng với kỳ vọng và độ lệch chuẩn của quần thể.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Tạo quần thể mô phỏng 10000 phần tử chiều cao từ 120-200cm
np.random.seed(42)  # Để có kết quả nhất quán
POP = np.random.normal(loc=170, scale=15, size=10000)
POP = np.clip(POP, 120, 200)  # Giới hạn trong khoảng 120-200cm

# a. Tính trung bình và độ lệch chuẩn
pop_mean = np.mean(POP)
pop_std = np.std(POP, ddof=0)  # ddof=0 cho quần thể
print(f"Chiều cao trung bình quần thể: {pop_mean:.2f} cm")
print(f"Độ lệch chuẩn quần thể: {pop_std:.2f} cm")

# b. Tỷ lệ người cao (>=180cm)
tall_ratio = np.sum(POP >= 180) / len(POP)
print(f"Tỷ lệ người cao (>=180cm): {tall_ratio:.3f} ({tall_ratio*100:.1f}%)")

# c. Vẽ histogram
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.hist(POP, bins=30, alpha=0.7, density=True, color='skyblue', edgecolor='black')
plt.title('Histogram Chiều cao Quần thể')
plt.xlabel('Chiều cao (cm)')
plt.ylabel('Mật độ')

# d. So sánh với phân phối chuẩn
plt.subplot(1, 2, 2)
plt.hist(POP, bins=30, alpha=0.7, density=True, color='skyblue', edgecolor='black', label='Dữ liệu')

# Vẽ đường cong phân phối chuẩn
x = np.linspace(120, 200, 1000)
y = stats.norm.pdf(x, loc=pop_mean, scale=pop_std)
plt.plot(x, y, 'r-', linewidth=2, label=f'Chuẩn N({pop_mean:.1f}, {pop_std:.1f}²)')
plt.title('So sánh với Phân phối Chuẩn')
plt.xlabel('Chiều cao (cm)')
plt.ylabel('Mật độ')
plt.legend()

plt.tight_layout()
plt.show()

print("\nNhận xét: Quần thể có phân phối gần với phân phối chuẩn")
```

### Bài 2: Lấy mẫu cỡ 20

Bạn hãy lấy một mẫu kích thước 20 phần từ. Tính:

a. Trung bình mẫu, và độ lệch chuẩn của mẫu

b. Tỷ lệ người cao của mẫu

c. Vẽ histogram của mẫu

d. Thử vẽ đồ thị hàm mật độ của phân phối chuẩn với tham số loc và scale bằng với giá trị trung bình và độ lệch chuẩn và chiều cao của mẫu. Bạn có nhận xét gì không?

```python
# Lấy mẫu cỡ 20
sample_20 = np.random.choice(POP, size=20, replace=False)

# a. Trung bình và độ lệch chuẩn mẫu
sample_mean_20 = np.mean(sample_20)
sample_std_20 = np.std(sample_20, ddof=1)  # ddof=1 cho mẫu
print(f"Trung bình mẫu (n=20): {sample_mean_20:.2f} cm")
print(f"Độ lệch chuẩn mẫu (n=20): {sample_std_20:.2f} cm")

# b. Tỷ lệ người cao trong mẫu
tall_ratio_20 = np.sum(sample_20 >= 180) / len(sample_20)
print(f"Tỷ lệ người cao trong mẫu: {tall_ratio_20:.3f} ({tall_ratio_20*100:.1f}%)")

# c & d. Vẽ histogram và so sánh với phân phối chuẩn
plt.figure(figsize=(12, 5))

# Histogram mẫu
plt.subplot(1, 2, 1)
plt.hist(sample_20, bins=8, alpha=0.7, color='lightgreen', edgecolor='black')
plt.title('Histogram Mẫu (n=20)')
plt.xlabel('Chiều cao (cm)')
plt.ylabel('Tần số')

# So sánh với phân phối chuẩn
plt.subplot(1, 2, 2)
plt.hist(sample_20, bins=8, alpha=0.7, density=True, color='lightgreen', edgecolor='black', label='Mẫu')

# Vẽ đường cong phân phối chuẩn từ tham số mẫu
x = np.linspace(min(sample_20)-10, max(sample_20)+10, 1000)
y = stats.norm.pdf(x, loc=sample_mean_20, scale=sample_std_20)
plt.plot(x, y, 'r-', linewidth=2, label=f'Chuẩn từ mẫu N({sample_mean_20:.1f}, {sample_std_20:.1f}²)')

# Vẽ đường cong phân phối chuẩn từ quần thể để so sánh
y_pop = stats.norm.pdf(x, loc=pop_mean, scale=pop_std)
plt.plot(x, y_pop, 'b--', linewidth=2, label=f'Chuẩn quần thể N({pop_mean:.1f}, {pop_std:.1f}²)')

plt.title('So sánh Mẫu với Phân phối Chuẩn')
plt.xlabel('Chiều cao (cm)')
plt.ylabel('Mật độ')
plt.legend()

plt.tight_layout()
plt.show()

print(f"\nNhận xét: Với mẫu nhỏ (n=20), ước lượng có thể khác khá nhiều so với quần thể")
print(f"Chênh lệch trung bình: {abs(sample_mean_20 - pop_mean):.2f} cm")
print(f"Chênh lệch độ lệch chuẩn: {abs(sample_std_20 - pop_std):.2f} cm")
```

### Bài 4: Khảo sát phân phối của trung bình mẫu

a. Thực hiện bài 2 100 lần, mỗi lần bạn tính được trung bình mẫu. Vẽ đồ thị histogram cho 100 trung bình mẫu bạn tính được.

b. Thực hiện bài 3 100 lần, mỗi lần bạn tính được trung bình mẫu. Vẽ đồ thị histogram cho 100 trung bình mẫu bạn tính được.

c. Vẽ histogram của hai câu a, b trong cùng một hình để so sánh. Theo bạn hình dạng của histogram thay đổi nói lên điều gì?

Gợi ý: tạo một danh sách rỗng để chứa các trung bình mẫu tính được, sử dụng hàm append() để thêm giá trị trung bình vào danh sách sau mỗi lần tính.

```python
# a. 100 lần lấy mẫu n=20
sample_means_20 = []
for i in range(100):
    sample = np.random.choice(POP, size=20, replace=False)
    sample_means_20.append(np.mean(sample))

# b. 100 lần lấy mẫu n=100  
sample_means_100 = []
for i in range(100):
    sample = np.random.choice(POP, size=100, replace=False)
    sample_means_100.append(np.mean(sample))

# Tính thống kê
print(f"Trung bình của các TB mẫu (n=20): {np.mean(sample_means_20):.2f}")
print(f"Độ lệch chuẩn của các TB mẫu (n=20): {np.std(sample_means_20):.2f}")
print(f"Trung bình của các TB mẫu (n=100): {np.mean(sample_means_100):.2f}")
print(f"Độ lệch chuẩn của các TB mẫu (n=100): {np.std(sample_means_100):.2f}")

# c. Vẽ histogram so sánh
plt.figure(figsize=(15, 5))

# Histogram riêng cho n=20
plt.subplot(1, 3, 1)
plt.hist(sample_means_20, bins=15, alpha=0.7, color='lightgreen', edgecolor='black')
plt.title('Phân phối TB mẫu (n=20)')
plt.xlabel('Trung bình mẫu')
plt.ylabel('Tần số')
plt.axvline(pop_mean, color='red', linestyle='--', label=f'TB quần thể = {pop_mean:.1f}')
plt.legend()

# Histogram riêng cho n=100
plt.subplot(1, 3, 2)
plt.hist(sample_means_100, bins=15, alpha=0.7, color='orange', edgecolor='black')
plt.title('Phân phối TB mẫu (n=100)')
plt.xlabel('Trung bình mẫu')
plt.ylabel('Tần số')
plt.axvline(pop_mean, color='red', linestyle='--', label=f'TB quần thể = {pop_mean:.1f}')
plt.legend()

# So sánh trực tiếp
plt.subplot(1, 3, 3)
plt.hist(sample_means_20, bins=15, alpha=0.5, color='lightgreen', label='n=20', density=True)
plt.hist(sample_means_100, bins=15, alpha=0.5, color='orange', label='n=100', density=True)
plt.title('So sánh Phân phối TB mẫu')
plt.xlabel('Trung bình mẫu')
plt.ylabel('Mật độ')
plt.axvline(pop_mean, color='red', linestyle='--', label=f'TB quần thể = {pop_mean:.1f}')
plt.legend()

plt.tight_layout()
plt.show()

print(f"\nNhận xét:")
print(f"- Cả hai phân phối đều tập trung quanh trung bình quần thể ({pop_mean:.2f})")
print(f"- Phân phối TB mẫu với n=100 có độ phân tán nhỏ hơn (ít biến động hơn)")
print(f"- Điều này chứng minh Định lý Giới hạn Trung tâm: TB mẫu có phân phối chuẩn")
print(f"- Khi n tăng, độ lệch chuẩn của TB mẫu giảm theo tỉ lệ 1/√n")

# Tính sai số chuẩn lý thuyết
se_20 = pop_std / np.sqrt(20)
se_100 = pop_std / np.sqrt(100)
print(f"\nSai số chuẩn lý thuyết:")
print(f"n=20: {se_20:.2f}, thực tế: {np.std(sample_means_20):.2f}")
print(f"n=100: {se_100:.2f}, thực tế: {np.std(sample_means_100):.2f}")
```

### Bài 5: Khảo sát tỷ lệ mẫu

a. Bạn hãy mô phỏng 1000 lần lấy mẫu có kích cỡ 100. Vẽ histogram tỷ lệ người cao của các mẫu

b. Bạn hãy mô phỏng 1000 lần lấy mẫu có kích cỡ 500. Vẽ histogram tỷ lệ người cao của các mẫu

```python
# a. 1000 lần lấy mẫu n=100
proportions_100 = []
for i in range(1000):
    sample = np.random.choice(POP, size=100, replace=False)
    prop = np.sum(sample >= 180) / len(sample)
    proportions_100.append(prop)

# b. 1000 lần lấy mẫu n=500
proportions_500 = []
for i in range(1000):
    sample = np.random.choice(POP, size=500, replace=False)
    prop = np.sum(sample >= 180) / len(sample)
    proportions_500.append(prop)

# Tính thống kê
print(f"Tỷ lệ cao thực tế trong quần thể: {tall_ratio:.3f}")
print(f"Trung bình tỷ lệ mẫu (n=100): {np.mean(proportions_100):.3f}")
print(f"Độ lệch chuẩn tỷ lệ mẫu (n=100): {np.std(proportions_100):.3f}")
print(f"Trung bình tỷ lệ mẫu (n=500): {np.mean(proportions_500):.3f}")
print(f"Độ lệch chuẩn tỷ lệ mẫu (n=500): {np.std(proportions_500):.3f}")

# Vẽ histogram so sánh
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.hist(proportions_100, bins=20, alpha=0.7, color='lightblue', edgecolor='black')
plt.title('Phân phối Tỷ lệ Mẫu (n=100)')
plt.xlabel('Tỷ lệ người cao')
plt.ylabel('Tần số')
plt.axvline(tall_ratio, color='red', linestyle='--', label=f'Tỷ lệ quần thể = {tall_ratio:.3f}')
plt.legend()

plt.subplot(1, 3, 2)
plt.hist(proportions_500, bins=20, alpha=0.7, color='lightcoral', edgecolor='black')
plt.title('Phân phối Tỷ lệ Mẫu (n=500)')
plt.xlabel('Tỷ lệ người cao')
plt.ylabel('Tần số')
plt.axvline(tall_ratio, color='red', linestyle='--', label=f'Tỷ lệ quần thể = {tall_ratio:.3f}')
plt.legend()

plt.subplot(1, 3, 3)
plt.hist(proportions_100, bins=20, alpha=0.5, color='lightblue', label='n=100', density=True)
plt.hist(proportions_500, bins=20, alpha=0.5, color='lightcoral', label='n=500', density=True)
plt.title('So sánh Phân phối Tỷ lệ Mẫu')
plt.xlabel('Tỷ lệ người cao')
plt.ylabel('Mật độ')
plt.axvline(tall_ratio, color='red', linestyle='--', label=f'Tỷ lệ quần thể = {tall_ratio:.3f}')
plt.legend()

plt.tight_layout()
plt.show()

print(f"\nNhận xét:")
print(f"- Tỷ lệ mẫu tập trung quanh tỷ lệ thực của quần thể")
print(f"- Với n=500, phân phối tỷ lệ mẫu ít biến động hơn so với n=100")
print(f"- Cả hai phân phối đều có dạng gần như phân phối chuẩn (CLT cho tỷ lệ)")

# Sai số chuẩn lý thuyết cho tỷ lệ
se_prop_100 = np.sqrt(tall_ratio * (1 - tall_ratio) / 100)
se_prop_500 = np.sqrt(tall_ratio * (1 - tall_ratio) / 500)
print(f"\nSai số chuẩn lý thuyết cho tỷ lệ:")
print(f"n=100: {se_prop_100:.3f}, thực tế: {np.std(proportions_100):.3f}")
print(f"n=500: {se_prop_500:.3f}, thực tế: {np.std(proportions_500):.3f}")
```

### Bài 6: Khảo sát phương sai mẫu

Bạn hãy mô phỏng 1000 lần lấy mẫu có kích cỡ 100. Vẽ histogram đại lượng phương sai của các mẫu (Phương sai bằng bình phương độ lệch chuẩn)

```python
# 1000 lần lấy mẫu n=100 để tính phương sai
sample_variances = []
for i in range(1000):
    sample = np.random.choice(POP, size=100, replace=False)
    var = np.var(sample, ddof=1)  # ddof=1 cho phương sai mẫu
    sample_variances.append(var)

# Tính thống kê
pop_variance = pop_std**2
sample_var_mean = np.mean(sample_variances)
sample_var_std = np.std(sample_variances)

print(f"Phương sai thực của quần thể: {pop_variance:.2f}")
print(f"Trung bình các phương sai mẫu: {sample_var_mean:.2f}")
print(f"Độ lệch chuẩn của các phương sai mẫu: {sample_var_std:.2f}")

# Vẽ histogram
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(sample_variances, bins=25, alpha=0.7, color='plum', edgecolor='black')
plt.title('Phân phối Phương sai Mẫu (n=100)')
plt.xlabel('Phương sai mẫu')
plt.ylabel('Tần số')
plt.axvline(pop_variance, color='red', linestyle='--', linewidth=2, label=f'Phương sai quần thể = {pop_variance:.1f}')
plt.axvline(sample_var_mean, color='blue', linestyle='-', linewidth=2, label=f'TB phương sai mẫu = {sample_var_mean:.1f}')
plt.legend()

# So sánh với phân phối Chi-square (lý thuyết)
plt.subplot(1, 2, 2)
plt.hist(sample_variances, bins=25, alpha=0.7, density=True, color='plum', edgecolor='black', label='Phương sai mẫu')

# Phân phối Chi-square có liên quan đến phương sai mẫu
# (n-1)S²/σ² ~ χ²(n-1)
# Vậy S² ~ σ²χ²(n-1)/(n-1)
df = 99  # n-1 = 100-1
chi2_samples = np.random.chisquare(df, 1000) * pop_variance / df
plt.hist(chi2_samples, bins=25, alpha=0.5, density=True, color='yellow', label='Chi-square lý thuyết')

plt.title('So sánh với Phân phối Chi-square')
plt.xlabel('Phương sai')
plt.ylabel('Mật độ')
plt.legend()

plt.tight_layout()
plt.show()

print(f"\nNhận xét:")
print(f"- Trung bình các phương sai mẫu gần bằng phương sai quần thể")
print(f"- Điều này chứng tỏ phương sai mẫu là ước lượng không thiên vị cho phương sai quần thể")
print(f"- Phân phối phương sai mẫu có dạng lệch phải, liên quan đến phân phối Chi-square")
print(f"- Sai số: {abs(sample_var_mean - pop_variance):.2f}")
```

### Bài 7:  Bài Tập Ước Lượng 

Khảo sát cân nặng (kg) của gà khi xuất chuồng, người ta cân một số con và có kết quả như sau:

Giả sử cân nặng của gà là biến ngẫu nhiên có phân phối chuẩn. Hãy lập khoảng tin cậy cho cân nặng trung bình của  khi xuất chuồng với độ tin cậy là 95%. Trong 2 trường hợp:

```python
# Dữ liệu cân nặng gà
data = np.array([2.1, 1.8, 2.0, 2.3, 1.7, 1.5, 2.0, 2.2, 1.8])

# Tính thống kê mô tả
n = len(data)
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)
confidence_level = 0.95
alpha = 1 - confidence_level

print(f"Dữ liệu: {data}")
print(f"Cỡ mẫu (n): {n}")
print(f"Trung bình mẫu: {sample_mean:.3f} kg")
print(f"Độ lệch chuẩn mẫu: {sample_std:.3f} kg")
print(f"Độ tin cậy: {confidence_level*100}%")

print(f"\n{'='*50}")
print("a) TRƯỜNG HỢP BIẾT σ = 0.3")
print(f"{'='*50}")

# a) Biết sigma = 0.3
sigma_known = 0.3
z_alpha_2 = stats.norm.ppf(1 - alpha/2)
margin_error_z = z_alpha_2 * sigma_known / np.sqrt(n)

lower_z = sample_mean - margin_error_z
upper_z = sample_mean + margin_error_z

print(f"Độ lệch chuẩn quần thể (σ): {sigma_known} kg")
print(f"Giá trị Z_{alpha/2}: {z_alpha_2:.3f}")
print(f"Sai số chuẩn: {sigma_known / np.sqrt(n):.3f}")
print(f"Sai số ước lượng: {margin_error_z:.3f}")
print(f"Khoảng tin cậy 95%: [{lower_z:.3f}, {upper_z:.3f}] kg")

print(f"\n{'='*50}")
print("b) TRƯỜNG HỢP KHÔNG BIẾT σ")
print(f"{'='*50}")

# b) Không biết sigma - sử dụng phân phối t
df = n - 1
t_alpha_2 = stats.t.ppf(1 - alpha/2, df)
margin_error_t = t_alpha_2 * sample_std / np.sqrt(n)

lower_t = sample_mean - margin_error_t
upper_t = sample_mean + margin_error_t

print(f"Bậc tự do (df): {df}")
print(f"Giá trị t_{alpha/2}: {t_alpha_2:.3f}")
print(f"Sai số chuẩn: {sample_std / np.sqrt(n):.3f}")
print(f"Sai số ước lượng: {margin_error_t:.3f}")
print(f"Khoảng tin cậy 95%: [{lower_t:.3f}, {upper_t:.3f}] kg")

# Vẽ minh họa
plt.figure(figsize=(14, 6))

# Biết sigma
plt.subplot(1, 2, 1)
x = np.linspace(sample_mean - 4*sigma_known/np.sqrt(n), sample_mean + 4*sigma_known/np.sqrt(n), 1000)
y = stats.norm.pdf(x, sample_mean, sigma_known/np.sqrt(n))
plt.plot(x, y, 'b-', linewidth=2, label='Phân phối chuẩn')
plt.axvline(lower_z, color='red', linestyle='--', label=f'KTC: [{lower_z:.3f}, {upper_z:.3f}]')
plt.axvline(upper_z, color='red', linestyle='--')
plt.axvline(sample_mean, color='green', linestyle='-', label=f'TB mẫu = {sample_mean:.3f}')
plt.fill_between(x, y, where=(x >= lower_z) & (x <= upper_z), alpha=0.3, color='red')
plt.title('Khoảng tin cậy (biết σ)')
plt.xlabel('Cân nặng (kg)')
plt.ylabel('Mật độ')
plt.legend()

# Không biết sigma
plt.subplot(1, 2, 2)
x = np.linspace(sample_mean - 4*sample_std/np.sqrt(n), sample_mean + 4*sample_std/np.sqrt(n), 1000)
y = stats.t.pdf(x, df, loc=sample_mean, scale=sample_std/np.sqrt(n))
plt.plot(x, y, 'orange', linewidth=2, label='Phân phối t')
plt.axvline(lower_t, color='red', linestyle='--', label=f'KTC: [{lower_t:.3f}, {upper_t:.3f}]')
plt.axvline(upper_t, color='red', linestyle='--')
plt.axvline(sample_mean, color='green', linestyle='-', label=f'TB mẫu = {sample_mean:.3f}')
plt.fill_between(x, y, where=(x >= lower_t) & (x <= upper_t), alpha=0.3, color='red')
plt.title('Khoảng tin cậy (không biết σ)')
plt.xlabel('Cân nặng (kg)')
plt.ylabel('Mật độ')
plt.legend()

plt.tight_layout()
plt.show()

print(f"\n{'='*50}")
print("KẾT LUẬN")
print(f"{'='*50}")
print(f"Với độ tin cậy 95%:")
print(f"- Biết σ = 0.3: Cân nặng TB quần thể trong khoảng [{lower_z:.3f}, {upper_z:.3f}] kg")
print(f"- Không biết σ: Cân nặng TB quần thể trong khoảng [{lower_t:.3f}, {upper_t:.3f}] kg")
print(f"\nNhận xét:")
print(f"- Khi không biết σ, khoảng tin cậy rộng hơn do sử dụng phân phối t")
print(f"- Độ rộng KTC biết σ: {upper_z - lower_z:.3f} kg")
print(f"- Độ rộng KTC không biết σ: {upper_t - lower_t:.3f} kg")
```
