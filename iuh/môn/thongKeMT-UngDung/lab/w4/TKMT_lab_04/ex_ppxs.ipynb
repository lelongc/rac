**Bài 01:** Một bài thi trắc nghiệm gồm 10 câu hỏi, mỗi câu có 4 phương án trả lời trong đó chỉ có một phương án đúng. Bạn hãy: 

a. Một sinh viên không học bài làm bài bằng cách chọn ngẫu nhiên một phương án cho mỗi câu hỏi. Bạn hãy dùng hàm random để in ra số câu đúng của bạn sinh viên trong 10 lần kiểm tra.

b. Giả sử mỗi câu đúng được 4 điểm, mỗi câu sai bị trừ 2 điểm. Tính xác suất để sinh viên này được 4 điểm

c. Gọi X là số câu trả lời đúng. Tính E(X) và Var(X)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Bài 01
print("BÀI 01:")

# a. Mô phỏng 10 lần kiểm tra
n = 10  # số câu hỏi
p = 1/4  # xác suất đúng mỗi câu

print("a. Số câu đúng trong 10 lần kiểm tra:")
for i in range(10):
    correct = np.random.binomial(n, p)
    print(f"Lần {i+1}: {correct} câu đúng")

# b. Tính xác suất được 4 điểm
print("\nb. Xác suất được 4 điểm:")
# Điểm = 4*đúng - 2*sai = 4*đúng - 2*(10-đúng) = 6*đúng - 20
# Để được 4 điểm: 6*đúng - 20 = 4 => đúng = 4 câu
prob = stats.binom.pmf(4, n, p)
print(f"P(được 4 điểm) = P(4 câu đúng) = {prob:.6f}")

# c. E(X) và Var(X)
print("\nc. E(X) và Var(X):")
E_X = n * p
Var_X = n * p * (1 - p)
print(f"E(X) = {E_X}")
print(f"Var(X) = {Var_X}")
```

**Bài 02:** Một trung tâm bưu điện nhận được 3 cuộc gọi điện thoại mỗi phút. Tính xác suất để trung tâm nhận được 1 cuộc gọi, 2 cuộc gọi, 3 cuộc gọi trong vòng một phút biết số cuộc gọi trong một phút có phân phối Poisson.

```python
# Bài 02
print("BÀI 02:")

# Phân phối Poisson với λ = 3
lam = 3

# Tính xác suất cho 1, 2, 3 cuộc gọi
for k in [1, 2, 3]:
    prob = stats.poisson.pmf(k, lam)
    print(f"P(X = {k}) = {prob:.6f}")
```

**Bài 03:** Trọng lượng (đơn vị gam) của một loại trái cây có phân phối chuẩn với µ = 500 (gam) gam và $σ^2 = 16 (gam^2)$. Trái cây thu hoạch được phân loại theo trọng lượng như sau:

Loại 1: trên 505 gam

Loại 2: từ 495 – 505 gam

Loại 3: dười 495 gam

a. Hãy tính tỷ lệ của mỗi loại.
b. Bạn hãy mô phỏng lấy mẫu  20 trái cây trên. Bạn hãy vẽ histogram trọng lượng trái cây và tính giá trị trung bình và độ lệch chuẩn trái cây từ mẫu
c. Bạn hãy mô phỏng lấy 100 trái cây trên. Bạn hãy vẽ histogram trọng lượng trái cây và tính giá trị trung bình và độ lệch chuẩn trái cây từ mẫu
d. Từ hai cách lấy mẫu ở câu b, c bạn có nhận xét gì không?

```python
# Bài 03
print("BÀI 03:")

mu = 500  # trung bình
sigma = 4  # độ lệch chuẩn (sqrt(16) = 4)

# a. Tỷ lệ của mỗi loại
print("a. Tỷ lệ từng loại:")
loai1 = 1 - stats.norm.cdf(505, mu, sigma)
loai2 = stats.norm.cdf(505, mu, sigma) - stats.norm.cdf(495, mu, sigma)
loai3 = stats.norm.cdf(495, mu, sigma)

print(f"Loại 1 (>505g): {loai1:.4f}")
print(f"Loại 2 (495-505g): {loai2:.4f}")
print(f"Loại 3 (<495g): {loai3:.4f}")

# b. Mẫu 20 trái cây
print("\nb. Mẫu 20 trái cây:")
sample20 = np.random.normal(mu, sigma, 20)
print(f"Trung bình: {np.mean(sample20):.2f}")
print(f"Độ lệch chuẩn: {np.std(sample20, ddof=1):.2f}")

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.hist(sample20, bins=5, alpha=0.7)
plt.title('Histogram mẫu 20')
plt.xlabel('Trọng lượng (g)')

# c. Mẫu 100 trái cây
print("\nc. Mẫu 100 trái cây:")
sample100 = np.random.normal(mu, sigma, 100)
print(f"Trung bình: {np.mean(sample100):.2f}")
print(f"Độ lệch chuẩn: {np.std(sample100, ddof=1):.2f}")

plt.subplot(1, 2, 2)
plt.hist(sample100, bins=10, alpha=0.7)
plt.title('Histogram mẫu 100')
plt.xlabel('Trọng lượng (g)')
plt.tight_layout()
plt.show()

# d. Nhận xét
print("\nd. Nhận xét:")
print("- Mẫu 100 gần với phân phối lý thuyết hơn mẫu 20")
print("- Mẫu lớn hơn cho ước lượng chính xác hơn")
```
