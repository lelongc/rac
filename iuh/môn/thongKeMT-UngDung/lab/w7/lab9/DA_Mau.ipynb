# Đề đại trà

```python
import numpy as np
from scipy import stats
from scipy.stats import norm
from scipy.stats import t
```

```python
# ---------------------
# Câu 1:
# ---------------------

city = [41.2, 42.7, 40.8, 43.1, 39.9, 42.3, 40.5, 43.7, 41.5, 42.9]
highway = [50.5, 52.1, 48.9, 49.3, 47.5, 51.0, 48.7, 50.2, 49.1, 50.0]
mixed = [44.3, 43.9, 45.6, 46.0, 44.8, 45.2, 46.1, 45.0, 44.6, 45.4]
```

```python
## (a)
highway_mean = np.mean(highway)
highway_std = np.std(highway)
se_highway = highway_std / np.sqrt(len(highway))
print("Câu 1(a):")
print("Trung bình mẫu:", round(highway_mean, 2))
print("Độ lệch chuẩn mẫu:", round(highway_std, 3))
print("Độ lệch chuẩn của trung bình mẫu:", round(se_highway, 4))
print()

## (b)
mixed_mean = np.mean(mixed)
mixed_std = np.std(mixed)
se_mixed = mixed_std / np.sqrt(len(mixed))
t_val = stats.t.ppf(1 - 0.001, df=len(mixed)-1)
ci_mixed = (mixed_mean - t_val * se_mixed, mixed_mean + t_val * se_mixed)
print("Câu 1(b):")
print("Trung bình mẫu:", round(mixed_mean, 2))
print("Độ lệch chuẩn mẫu:", round(mixed_std, 3))
print("Độ lệch chuẩn của trung bình mẫu:", round(se_mixed, 4))
print("99% CI:", tuple(float(round(val, 3)) for val in ci_mixed))
print()

## (c)
success = sum([1 for x in mixed if x >= 45])
p_hat = success / len(mixed)
se_p = np.sqrt(p_hat * (1 - p_hat) / len(mixed))
z = stats.norm.ppf(0.95)
ci_p = (p_hat - z * se_p, p_hat + z * se_p)
print("Câu 1(c):")
print("Tỷ lệ mẫu:", round(p_hat, 3))
print("SE = :", round(se_p, 4))
print("95% CI:", tuple(float(round(val, 3)) for val in ci_p))
print()

```

```python
# ---------------------
# Câu 2:
# ---------------------

## a)
# Dữ liệu mẫu
x1 = np.array([110, 125, 118, 130, 122, 115, 128, 117, 120, 123, 127, 124, 129, 119, 121])
mu0 = 120      # Giả thuyết H0: μ = 120
sigma = 6      # Đã biết phương sai => kiểm định Z
n1 = len(x1)

# Trung bình mẫu và giá trị Z
xbar1 = np.mean(x1)
z1 = (xbar1 - mu0) / (sigma / np.sqrt(n1))

# Mức ý nghĩa
alpha = 0.05
z_critical = norm.ppf(1 - alpha / 2)
# In kết quả
print("Câu 2(a):")
print("Trung bình mẫu:", round(xbar1, 2))
print("Z statistic:", round(z1, 3))
print("Z critical (alpha = 0.05): ±", round(z_critical, 3))

# Kết luận
if abs(z1) > z_critical:
    print("Bác bỏ H0: Có bằng chứng cho thấy thời gian sử dụng ≠ 120 phút.")
else:
    print("Không bác bỏ H0: Không có bằng chứng cho thấy thời gian sử dụng khác 120 phút.")

## b)
# Dữ liệu chi tiêu
x2 = np.array([520, 480, 495, 510, 470, 530, 490, 485, 505, 515])
mu0_b = 500  # Giả thuyết H0: μ = 500

# Thống kê mẫu
xbar2 = np.mean(x2)
s2 = np.std(x2, ddof=1)
n2 = len(x2)
t_stat = (xbar2 - mu0_b) / (s2 / np.sqrt(n2))

# Mức ý nghĩa và giá trị tới hạn
alpha = 0.01
df = n2 - 1
t_critical = t.ppf(1 - alpha/2, df)

# In kết quả
print("\nCâu 2(b):")
print("Trung bình mẫu:", round(xbar2, 2))
print("Độ lệch chuẩn mẫu:", round(s2, 2))
print("t statistic:", round(t_stat, 3))
print("t critical (df=9, alpha=0.01): ±", round(t_critical, 3))

# Kết luận
if abs(t_stat) > t_critical:
    print("Bác bỏ H0: Có bằng chứng cho thấy chi tiêu trung bình khác 500 nghìn.")
else:
    print("Không bác bỏ H0: Không có bằng chứng cho thấy chi tiêu trung bình khác 500 nghìn.")

## c)
# Dữ liệu khảo sát
n_2c = 100
successes_2c = 68
p_hat_2c = successes_2c / n_2c
p0_2c = 0.6

# Độ lệch chuẩn và thống kê Z
se_2c = np.sqrt(p0_2c * (1 - p0_2c) / n_2c)
z_stat_2c = (p_hat_2c - p0_2c) / se_2c
z_critical_2c = stats.norm.ppf(1 - 0.05)  # kiểm định một phía

# In kết quả
print("\nCâu 2(c):")
print("Tỷ lệ mẫu:", round(p_hat_2c, 2))
print("Z statistic:", round(z_stat_2c, 3))
print("Z critical (một phía, alpha=0.05):", round(z_critical_2c, 3))

# Kết luận
if z_stat_2c > z_critical_2c:
    print("Bác bỏ H0: Có bằng chứng cho thấy tỷ lệ vượt quá 60%.")
else:
    print("Không bác bỏ H0: Không đủ bằng chứng cho thấy tỷ lệ vượt quá 60%.")
```
