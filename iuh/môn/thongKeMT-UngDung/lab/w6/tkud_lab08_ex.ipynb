### Bài 1: Tạo quần thể mô phỏng

Bạn hãy tạo một mẫu gồm 10000 phần tử mô phỏng chiều cao của nam thanh niên có giá trị từ 120cm - 200cm. Bạn hãy lưu kết quả vào biến **POP**.

Tính:
a. Chiều cao trung bình (kỳ vọng) của quần thể và độ lệch chuẩn về chiều cao của quần thể.

b. Tính tỷ lệ người cao trong quần thể, biết rằng thanh niên có chiều cao từ 180 trở lên được gọi là cao.

c. Vẽ histogram về chiều cao của quần thể. Theo bạn quần thể có phân phối chuẩn hay không?

d. Thử vẽ histogram và đồ thị hàm mật độ của phân phối chuẩn sử dụng tham số loc và scale bằng với kỳ vọng và độ lệch chuẩn của quần thể.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
import math

# Bài 1: Tạo quần thể mô phỏng
np.random.seed(42)  # Để kết quả tái lập được
POP = np.random.uniform(120, 200, size=10000)  # Tạo quần thể 10000 phần tử từ 120-200cm

# a. Tính chiều cao trung bình và độ lệch chuẩn
pop_mean = np.mean(POP)
pop_std = np.std(POP)
print(f"a. Chiều cao trung bình của quần thể: {pop_mean:.2f} cm")
print(f"   Độ lệch chuẩn của quần thể: {pop_std:.2f} cm")

# b. Tính tỷ lệ người cao (>=180cm)
tall_people = np.sum(POP >= 180)
tall_ratio = tall_people / len(POP)
print(f"b. Số người cao (>=180cm): {tall_people}")
print(f"   Tỷ lệ người cao trong quần thể: {tall_ratio:.4f} ({tall_ratio*100:.2f}%)")

# c. Vẽ histogram về chiều cao của quần thể
plt.figure(figsize=(10, 6))
plt.hist(POP, bins=30, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(x=pop_mean, color='red', linestyle='--', label=f'Trung bình: {pop_mean:.2f}cm')
plt.title('Histogram chiều cao của quần thể')
plt.xlabel('Chiều cao (cm)')
plt.ylabel('Tần số')
plt.legend()
plt.grid(axis='y', alpha=0.75)
plt.show()

# d. Vẽ histogram và đồ thị hàm mật độ của phân phối chuẩn
x = np.linspace(120, 200, 1000)
pdf = stats.norm.pdf(x, loc=pop_mean, scale=pop_std)

plt.figure(figsize=(10, 6))
plt.hist(POP, bins=30, density=True, color='skyblue', edgecolor='black', alpha=0.7, label='Histogram')
plt.plot(x, pdf, 'r-', linewidth=2, label=f'Phân phối chuẩn\n(μ={pop_mean:.2f}, σ={pop_std:.2f})')
plt.title('So sánh histogram với phân phối chuẩn')
plt.xlabel('Chiều cao (cm)')
plt.ylabel('Mật độ xác suất')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print("c-d. Nhận xét: Histogram của quần thể có dạng phân phối đều, không phải phân phối chuẩn.")
print("     Điều này phù hợp vì dữ liệu được tạo từ phân phối đều uniform(120, 200).")
```

### Bài 2: Lấy mẫu cỡ 20

Bạn hãy lấy một mẫu kích thước 20 phần từ. Tính:

a. Trung bình mẫu, và độ lệch chuẩn của mẫu

b. Tỷ lệ người cao của mẫu

c. Vẽ histogram của mẫu

d. Thử vẽ đồ thị hàm mật độ của phân phối chuẩn với tham số loc và scale bằng với giá trị trung bình và độ lệch chuẩn và chiều cao của mẫu. Bạn có nhận xét gì không?

### Bài 4: Khảo sát phân phối của trung bình mẫu

a. Thực hiện bài 2 100 lần, mỗi lần bạn tính được trung bình mẫu. Vẽ đồ thị histogram cho 100 trung bình mẫu bạn tính được.

b. Thực hiện bài 3 100 lần, mỗi lần bạn tính được trung bình mẫu. Vẽ đồ thị histogram cho 100 trung bình mẫu bạn tính được.

c. Vẽ histogram của hai câu a, b trong cùng một hình để so sánh. Theo bạn hình dạng của histogram thay đổi nói lên điều gì?

Gợi ý: tạo một danh sách rỗng để chứa các trung bình mẫu tính được, sử dụng hàm append() để thêm giá trị trung bình vào danh sách sau mỗi lần tính.

```python
# Bài 4: Khảo sát phân phối của trung bình mẫu
np.random.seed(42)

# a. Thực hiện lấy mẫu cỡ 20, 100 lần và tính trung bình mẫu
sample_means_20 = []
for _ in range(100):
    sample = np.random.choice(POP, size=20)
    sample_means_20.append(np.mean(sample))

# b. Thực hiện lấy mẫu cỡ 100, 100 lần và tính trung bình mẫu
sample_means_100 = []
for _ in range(100):
    sample = np.random.choice(POP, size=100)
    sample_means_100.append(np.mean(sample))

# Vẽ histogram cho 100 trung bình mẫu cỡ 20
plt.figure(figsize=(10, 6))
plt.hist(sample_means_20, bins=15, color='lightblue', edgecolor='black', alpha=0.7)
plt.axvline(x=pop_mean, color='red', linestyle='--', label=f'TB quần thể: {pop_mean:.2f}cm')
plt.title('Histogram của 100 trung bình mẫu cỡ 20')
plt.xlabel('Trung bình mẫu')
plt.ylabel('Tần số')
plt.legend()
plt.grid(axis='y', alpha=0.75)
plt.show()

# Vẽ histogram cho 100 trung bình mẫu cỡ 100
plt.figure(figsize=(10, 6))
plt.hist(sample_means_100, bins=15, color='lightgreen', edgecolor='black', alpha=0.7)
plt.axvline(x=pop_mean, color='red', linestyle='--', label=f'TB quần thể: {pop_mean:.2f}cm')
plt.title('Histogram của 100 trung bình mẫu cỡ 100')
plt.xlabel('Trung bình mẫu')
plt.ylabel('Tần số')
plt.legend()
plt.grid(axis='y', alpha=0.75)
plt.show()

# c. So sánh hai histogram
plt.figure(figsize=(12, 6))

# Tính thống kê cho các mẫu
mean_of_means_20 = np.mean(sample_means_20)
std_of_means_20 = np.std(sample_means_20, ddof=1)
mean_of_means_100 = np.mean(sample_means_100)
std_of_means_100 = np.std(sample_means_100, ddof=1)

# Vẽ histogram trên cùng một biểu đồ
plt.hist(sample_means_20, bins=15, alpha=0.5, color='blue', label=f'n=20, μ={mean_of_means_20:.2f}, σ={std_of_means_20:.2f}')
plt.hist(sample_means_100, bins=15, alpha=0.5, color='green', label=f'n=100, μ={mean_of_means_100:.2f}, σ={std_of_means_100:.2f}')
plt.axvline(x=pop_mean, color='red', linestyle='--', label=f'TB quần thể: {pop_mean:.2f}')
plt.title('So sánh phân phối của trung bình mẫu cỡ 20 và 100')
plt.xlabel('Trung bình mẫu')
plt.ylabel('Tần số')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print("c. Nhận xét về sự thay đổi của histogram:")
print("   1. Phân phối của trung bình mẫu đều xấp xỉ phân phối chuẩn (Định lý giới hạn trung tâm)")
print("   2. Cả hai phân phối đều tập trung quanh giá trị trung bình quần thể")
print("   3. Độ phân tán của phân phối trung bình mẫu cỡ 100 nhỏ hơn nhiều so với cỡ 20")
print("   4. Khi kích thước mẫu tăng, phương sai của trung bình mẫu giảm")
print(f"   5. Tỷ lệ độ lệch chuẩn: σ(n=20)/σ(n=100) ≈ {std_of_means_20/std_of_means_100:.2f} ≈ √(100/20) = {math.sqrt(100/20):.2f}")
```

### Bài 5: Khảo sát tỷ lệ mẫu

a. Bạn hãy mô phỏng 1000 lần lấy mẫu có kích cỡ 100. Vẽ histogram tỷ lệ người cao của các mẫu

b. Bạn hãy mô phỏng 1000 lần lấy mẫu có kích cỡ 500. Vẽ histogram tỷ lệ người cao của các mẫu

```python
# Bài 5: Khảo sát tỷ lệ mẫu
np.random.seed(42)

# Tính tỷ lệ người cao thực tế trong quần thể
pop_tall_ratio = np.mean(POP >= 180)
print(f"Tỷ lệ người cao trong quần thể: {pop_tall_ratio:.4f} ({pop_tall_ratio*100:.2f}%)")

# a. Mô phỏng 1000 lần lấy mẫu cỡ 100
sample_ratios_100 = []
for _ in range(1000):
    sample = np.random.choice(POP, size=100)
    tall_ratio = np.mean(sample >= 180)
    sample_ratios_100.append(tall_ratio)

# Vẽ histogram tỷ lệ người cao của các mẫu cỡ 100
plt.figure(figsize=(10, 6))
plt.hist(sample_ratios_100, bins=30, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(x=pop_tall_ratio, color='red', linestyle='--', 
            label=f'Tỷ lệ quần thể: {pop_tall_ratio:.4f}')

# Thêm đường cong phân phối chuẩn
x = np.linspace(min(sample_ratios_100), max(sample_ratios_100), 1000)
mean_ratio_100 = np.mean(sample_ratios_100)
std_ratio_100 = np.std(sample_ratios_100, ddof=1)
pdf_100 = stats.norm.pdf(x, loc=mean_ratio_100, scale=std_ratio_100)
plt.plot(x, pdf_100*len(sample_ratios_100)*(max(sample_ratios_100)-min(sample_ratios_100))/30, 
         'g-', linewidth=2, label=f'Phân phối chuẩn\n(μ={mean_ratio_100:.4f}, σ={std_ratio_100:.4f})')

plt.title('Histogram tỷ lệ người cao của 1000 mẫu cỡ 100')
plt.xlabel('Tỷ lệ người cao')
plt.ylabel('Tần số')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# b. Mô phỏng 1000 lần lấy mẫu cỡ 500
sample_ratios_500 = []
for _ in range(1000):
    sample = np.random.choice(POP, size=500)
    tall_ratio = np.mean(sample >= 180)
    sample_ratios_500.append(tall_ratio)

# Vẽ histogram tỷ lệ người cao của các mẫu cỡ 500
plt.figure(figsize=(10, 6))
plt.hist(sample_ratios_500, bins=30, color='lightgreen', edgecolor='black', alpha=0.7)
plt.axvline(x=pop_tall_ratio, color='red', linestyle='--', 
            label=f'Tỷ lệ quần thể: {pop_tall_ratio:.4f}')

# Thêm đường cong phân phối chuẩn
x = np.linspace(min(sample_ratios_500), max(sample_ratios_500), 1000)
mean_ratio_500 = np.mean(sample_ratios_500)
std_ratio_500 = np.std(sample_ratios_500, ddof=1)
pdf_500 = stats.norm.pdf(x, loc=mean_ratio_500, scale=std_ratio_500)
plt.plot(x, pdf_500*len(sample_ratios_500)*(max(sample_ratios_500)-min(sample_ratios_500))/30, 
         'g-', linewidth=2, label=f'Phân phối chuẩn\n(μ={mean_ratio_500:.4f}, σ={std_ratio_500:.4f})')

plt.title('Histogram tỷ lệ người cao của 1000 mẫu cỡ 500')
plt.xlabel('Tỷ lệ người cao')
plt.ylabel('Tần số')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# So sánh phương sai của hai phân phối
print(f"Độ lệch chuẩn của tỷ lệ mẫu cỡ 100: {std_ratio_100:.6f}")
print(f"Độ lệch chuẩn của tỷ lệ mẫu cỡ 500: {std_ratio_500:.6f}")
print(f"Tỷ lệ: {std_ratio_100/std_ratio_500:.2f} ≈ √(500/100) = {math.sqrt(500/100):.2f}")
print("\nNhận xét:")
print("1. Cả hai phân phối đều xấp xỉ phân phối chuẩn, tập trung quanh tỷ lệ thực của quần thể")
print("2. Phân phối của tỷ lệ mẫu cỡ 500 có độ phân tán nhỏ hơn nhiều so với cỡ 100")
print("3. Độ lệch chuẩn của tỷ lệ mẫu tỷ lệ nghịch với căn bậc hai của kích thước mẫu")
```

### Bài 6: Khảo sát phương sai mẫu

Bạn hãy mô phỏng 1000 lần lấy mẫu có kích cỡ 100. Vẽ histogram đại lượng phương sai của các mẫu (Phương sai bằng bình phương độ lệch chuẩn)

```python
# Bài 6: Khảo sát phương sai mẫu
np.random.seed(42)

# Tính phương sai của quần thể
pop_var = np.var(POP)
print(f"Phương sai của quần thể: {pop_var:.4f}")

# Mô phỏng 1000 lần lấy mẫu có kích cỡ 100
sample_vars = []
for _ in range(1000):
    sample = np.random.choice(POP, size=100)
    # Tính phương sai mẫu
    sample_var = np.var(sample, ddof=1)  # ddof=1 để tính phương sai mẫu
    sample_vars.append(sample_var)

# Vẽ histogram phương sai của các mẫu
plt.figure(figsize=(10, 6))
plt.hist(sample_vars, bins=30, color='lightcoral', edgecolor='black', alpha=0.7)
plt.axvline(x=pop_var, color='red', linestyle='--', 
            label=f'Phương sai quần thể: {pop_var:.4f}')

# Tính thống kê của phương sai mẫu
mean_var = np.mean(sample_vars)
std_var = np.std(sample_vars, ddof=1)

# Thêm đường phân phối lý thuyết (chi-squared)
# Trong lý thuyết, phương sai mẫu * (n-1)/σ² có phân phối chi-squared với n-1 bậc tự do
# Nhưng ở đây ta chỉ vẽ đường cong Gaussian tương ứng
x = np.linspace(min(sample_vars), max(sample_vars), 1000)
pdf = stats.norm.pdf(x, loc=mean_var, scale=std_var)
plt.plot(x, pdf*len(sample_vars)*(max(sample_vars)-min(sample_vars))/30, 
         'g-', linewidth=2, label=f'Phân phối xấp xỉ\n(μ={mean_var:.4f}, σ={std_var:.4f})')

plt.title('Histogram phương sai của 1000 mẫu cỡ 100')
plt.xlabel('Phương sai mẫu')
plt.ylabel('Tần số')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print(f"Giá trị trung bình của phương sai mẫu: {mean_var:.4f}")
print(f"Phương sai của quần thể: {pop_var:.4f}")
print(f"Tỷ lệ E[S²]/σ²: {mean_var/pop_var:.4f}")  # Nên xấp xỉ 1
print("\nNhận xét:")
print("1. Phương sai mẫu là một ước lượng không chệch của phương sai quần thể")
print("2. Phân phối của phương sai mẫu không đối xứng, có độ lệch dương")
print("3. Phân phối của phương sai mẫu có dạng xấp xỉ phân phối chi-squared")
```

### Bài 7:  Bài Tập Ước Lượng

Khảo sát cân nặng (kg) của gà khi xuất chuồng, người ta cân một số con và có kết quả như sau:

> 2.1, 1.8, 2.0, 2.3, 1.7, 1.5, 2.0, 2.2, 1.8

Giả sử cân nặng của gà là biến ngẫu nhiên có phân phối chuẩn. Hãy lập khoảng tin cậy cho cân nặng trung bình của  khi xuất chuồng với độ tin cậy là 95%. Trong 2 trường hợp:

a, Biết $\sigma = 0.3$

b, Không biết $\sigma$


```python
# Bài 7: Bài Tập Ước Lượng
# Dữ liệu cân nặng của gà khi xuất chuồng (kg)
weights = np.array([2.1, 1.8, 2.0, 2.3, 1.7, 1.5, 2.0, 2.2, 1.8])

# Tính các thống kê cơ bản
n = len(weights)
xbar = np.mean(weights)
s = np.std(weights, ddof=1)  # ddof=1 để tính độ lệch chuẩn mẫu
print(f"Số quan sát: n = {n}")
print(f"Trung bình mẫu: xbar = {xbar:.4f} kg")
print(f"Độ lệch chuẩn mẫu: s = {s:.4f} kg")

# a. Trường hợp biết σ = 0.3
sigma = 0.3
alpha = 0.05  # độ tin cậy 95%
z = stats.norm.ppf(1 - alpha/2)  # z_{1-α/2}
margin_error = z * sigma / np.sqrt(n)
ci_a_lower = xbar - margin_error
ci_a_upper = xbar + margin_error
print("\na. Trường hợp biết σ = 0.3:")
print(f"   z_{1-alpha/2} = {z:.4f}")
print(f"   Sai số biên = {margin_error:.4f}")
print(f"   Khoảng tin cậy 95%: [{ci_a_lower:.4f}, {ci_a_upper:.4f}]")

# b. Trường hợp không biết σ
alpha = 0.05  # độ tin cậy 95%
t = stats.t.ppf(1 - alpha/2, df=n-1)  # t_{1-α/2, n-1}
margin_error = t * s / np.sqrt(n)
ci_b_lower = xbar - margin_error
ci_b_upper = xbar + margin_error
print("\nb. Trường hợp không biết σ:")
print(f"   t_{1-alpha/2, n-1} = {t:.4f}")
print(f"   Sai số biên = {margin_error:.4f}")
print(f"   Khoảng tin cậy 95%: [{ci_b_lower:.4f}, {ci_b_upper:.4f}]")

# Minh họa khoảng tin cậy
plt.figure(figsize=(10, 6))
plt.scatter(weights, np.zeros_like(weights), color='blue', s=100, alpha=0.6, label='Mẫu quan sát')
plt.axvline(x=xbar, color='red', linewidth=2, label=f'Trung bình mẫu: {xbar:.4f}')
plt.axvspan(ci_a_lower, ci_a_upper, alpha=0.3, color='green', label=f'KTC 95% (σ đã biết): [{ci_a_lower:.4f}, {ci_a_upper:.4f}]')
plt.axvspan(ci_b_lower, ci_b_upper, alpha=0.3, color='orange', label=f'KTC 95% (σ chưa biết): [{ci_b_lower:.4f}, {ci_b_upper:.4f}]')
plt.axhline(y=0, color='black', linestyle='-', alpha=0.3)
plt.title('Khoảng tin cậy 95% cho cân nặng trung bình của gà')
plt.xlabel('Cân nặng (kg)')
plt.yticks([])
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print("\nNhận xét:")
print("1. Khi không biết σ, phải sử dụng phân phối t và độ lệch chuẩn mẫu s")
print("2. Khoảng tin cậy khi không biết σ rộng hơn so với khi biết σ")
print("3. Với cùng độ tin cậy 95%, khoảng tin cậy với t rộng hơn khoảng tin cậy với z")
```

Một khảo sát 600 người tiêu dùng cho biết **168 người** nói rằng họ **sẵn sàng mua hàng online thay vì tại cửa hàng**.

```python
# Bài 10: Ước lượng khoảng tin cậy cho tỷ lệ
n = 600
x = 168
p_hat = x / n

print(f"Số người được khảo sát: n = {n}")
print(f"Số người sẵn sàng mua online: x = {x}")
print(f"Tỷ lệ mẫu: p_hat = {p_hat:.4f} ({p_hat*100:.2f}%)")

# Tính khoảng tin cậy 99%
alpha = 0.01
z = stats.norm.ppf(1 - alpha/2)
margin_error = z * np.sqrt(p_hat * (1 - p_hat) / n)
ci_lower = max(0, p_hat - margin_error)
ci_upper = min(1, p_hat + margin_error)

print(f"\nKhoảng tin cậy 99% cho tỷ lệ thật sự: [{ci_lower:.4f}, {ci_upper:.4f}]")
print(f"Khoảng tin cậy 99% cho tỷ lệ thật sự: [{ci_lower*100:.2f}%, {ci_upper*100:.2f}%]")
```

# Bài 11

Một hãng thực phẩm quảng cáo rằng **hàm lượng muối trung bình** trong một gói snack của họ là **1.8 gram**.  
Một nhóm kiểm nghiệm độc lập lấy **30 gói ngẫu nhiên** và đo được (gram):

1.9, 1.7, 1.8, 1.6, 1.9, 2.0, 1.5, 1.7, 1.8, 1.6,  
1.9, 1.8, 2.1, 1.7, 1.6, 1.8, 2.0, 1.9, 1.8, 1.5,  
1.7, 1.8, 2.0, 1.9, 1.6, 1.7, 2.1, 1.9, 1.8, 1.7  

**Yêu cầu:**  
Kiểm định ở mức ý nghĩa **5%** xem **hàm lượng muối trung bình có khác 1.8 gram hay không**.


```python
# Bài 11: Kiểm định trung bình (1 mẫu)
salt_data = np.array([1.9, 1.7, 1.8, 1.6, 1.9, 2.0, 1.5, 1.7, 1.8, 1.6,
                      1.9, 1.8, 2.1, 1.7, 1.6, 1.8, 2.0, 1.9, 1.8, 1.5,
                      1.7, 1.8, 2.0, 1.9, 1.6, 1.7, 2.1, 1.9, 1.8, 1.7])

mu0 = 1.8  # Giá trị quảng cáo
n = len(salt_data)
xbar = np.mean(salt_data)
s = np.std(salt_data, ddof=1)

print(f"Lượng muối trung bình quảng cáo: μ₀ = {mu0} gram")
print(f"Số quan sát n = {n}")
print(f"Trung bình mẫu = {xbar:.4f} gram")
print(f"Độ lệch chuẩn mẫu = {s:.4f} gram")

# Thiết lập giả thuyết
print("\nGiả thuyết:")
print("H₀: μ = 1.8 (lượng muối trung bình đúng như quảng cáo)")
print("H₁: μ ≠ 1.8 (lượng muối trung bình khác với quảng cáo)")

# Tính thống kê kiểm định
alpha = 0.05
t_stat = (xbar - mu0) / (s / np.sqrt(n))
p_value = 2 * (1 - stats.t.cdf(abs(t_stat), df=n-1))  # Kiểm định hai phía

print(f"\nGiá trị thống kê t = {t_stat:.4f}")
print(f"p-value = {p_value:.4f}")
print(f"Mức ý nghĩa α = {alpha}")

# Kết luận
if p_value < alpha:
    conclusion = "Bác bỏ H₀"
    evidence = "Có bằng chứng cho thấy lượng muối trung bình khác với 1.8 gram"
else:
    conclusion = "Chấp nhận H₀"
    evidence = "Chưa có bằng chứng cho thấy lượng muối trung bình khác với 1.8 gram"
    
print(f"\nKết luận: {conclusion}")
print(f"Giải thích: {evidence}")
```

# Bài 12

Một công ty tuyên bố rằng **85% khách hàng hài lòng** với dịch vụ giao hàng.  
Một khảo sát độc lập với **400 khách hàng** cho thấy có **320 người hài lòng**.

**Yêu cầu:**  
Kiểm định ở mức ý nghĩa **5%** xem **tỷ lệ thật sự có nhỏ hơn 85% hay không**.


```python
# Bài 12: Kiểm định tỷ lệ (1 mẫu)
n = 400
x = 320
p_hat = x / n
p0 = 0.85  # tỷ lệ quảng cáo

print(f"Tỷ lệ hài lòng theo công ty tuyên bố: p₀ = {p0:.2f} ({p0*100}%)")
print(f"Số người được khảo sát: n = {n}")
print(f"Số người hài lòng: x = {x}")
print(f"Tỷ lệ mẫu: p_hat = {p_hat:.4f} ({p_hat*100:.2f}%)")

# Thiết lập giả thuyết
print("\nGiả thuyết:")
print(f"H₀: p = {p0} (tỷ lệ hài lòng đúng như tuyên bố)")
print(f"H₁: p < {p0} (tỷ lệ hài lòng thấp hơn tuyên bố)")

# Tính thống kê kiểm định
alpha = 0.05
z_stat = (p_hat - p0) / np.sqrt(p0 * (1 - p0) / n)
p_value = stats.norm.cdf(z_stat)  # Kiểm định một phía p < p0

print(f"\nGiá trị thống kê z = {z_stat:.4f}")
print(f"p-value = {p_value:.4f}")
print(f"Mức ý nghĩa α = {alpha}")

# Kết luận
if p_value < alpha:
    conclusion = "Bác bỏ H₀"
    evidence = "Có bằng chứng cho thấy tỷ lệ hài lòng thực tế nhỏ hơn 85%"
else:
    conclusion = "Chấp nhận H₀"
    evidence = "Chưa có bằng chứng cho thấy tỷ lệ hài lòng thực tế nhỏ hơn 85%"
    
print(f"\nKết luận: {conclusion}")
print(f"Giải thích: {evidence}")
```
