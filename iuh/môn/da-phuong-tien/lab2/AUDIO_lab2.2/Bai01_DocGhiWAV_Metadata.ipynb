# Bài 1: Đọc/Ghi WAV và Metadata

**Mục tiêu:**
- Mở tệp WAV, đọc số kênh, độ sâu bit, tần số lấy mẫu, số frame, thời lượng.
- Ghi một tệp WAV đơn giản từ mảng NumPy.

**Thư viện sử dụng:** `wave`, `scipy.io.wavfile`, `numpy`.  
**Lưu ý:** Không cần `librosa`.

```python
#@title Cài đặt (nếu cần)
# !pip install scipy
```

```python
import wave
import contextlib

filename = "sample-6s.wav"  # Thay bằng đường dẫn của bạn

with contextlib.closing(wave.open(filename, 'rb')) as audio:
    channels = audio.getnchannels()
    sample_width = audio.getsampwidth()
    framerate = audio.getframerate()
    nframes = audio.getnframes()
    duration = nframes / float(framerate)
    print("Số kênh:", channels)
    print("Kích thước mẫu (bytes):", sample_width)
    print("Tần số lấy mẫu (Hz):", framerate)
    print("Số frame:", nframes)
    print("Thời lượng (giây):", duration)
```

### Đọc nhanh bằng SciPy

```python
from scipy.io import wavfile
rate, data = wavfile.read(filename)
print("rate:", rate, "shape:", data.shape, "dtype:", data.dtype)
```

### Ghi một tệp WAV từ NumPy

```python
import numpy as np
from scipy.io.wavfile import write

fs = 16000  # Hz
t = np.linspace(0, 1.0, int(fs), endpoint=False)
tone = (0.2*np.sin(2*np.pi*440*t)).astype(np.float32)  # A4 = 440 Hz

# Chuẩn hóa về int16 để ghi WAV 16-bit
tone_int16 = (tone * 32767).astype(np.int16)
write("tone_440Hz.wav", fs, tone_int16)
print("Đã ghi tone_440Hz.wav")
```
