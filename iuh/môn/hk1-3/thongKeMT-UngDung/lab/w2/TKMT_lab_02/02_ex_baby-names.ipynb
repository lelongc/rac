Dữ Liệu `US_BaBy_Names.csv` chứa thông tin của các em bé sinh từ 2004 - 2014 tại Mỹ. Dựa vào tập dữ liệu trên, bạn hãy thực hiện các bài tập sau:

## TÌM HIỂU DỮ LIỆU

### 01. Đọc dữ liệu

```python
import pandas as pd

# Đọc dữ liệu từ file CSV
df = pd.read_csv('US_Baby_Names.csv')
```

### 02. Hiển thị 10 dòng dữ liệu bất kỳ

```python
# Hiển thị 10 dòng dữ liệu bất kỳ
df.sample(10)
```

### 03. Cho biết kích thước dữ liệu

```python
#  kích thước dữ liệu
print(f"Kích thước dữ liệu - Số dòng và cột: {df.shape}")
print(f"Số dòng: {df.shape[0]}")
print(f"Số cột: {df.shape[1]}")
```

### 04. Xóa 2 cột 'Unnamed: 0' and 'Id'

```python
# Xóa 2 cột 'Unnamed: 0' and 'Id'
df = df.drop(['Unnamed: 0', 'Id'], axis=1)
print("Đã xóa 2 cột 'Unnamed: 0' và 'Id'")
print(f"Kích thước dữ liệu sau khi xóa: {df.shape}")
df.head()
```

### 05. Cho biết số lượng bé trai, bé gái?

```python
# Đếm số lượng bé trai và bé gái
gender_counts = df.groupby('Gender')['Count'].sum()
print("Số lượng bé trai và bé gái:")
print(f"Bé gái (F): {gender_counts['F']:,}")
print(f"Bé trai (M): {gender_counts['M']:,}")
print(f"Tổng cộng: {gender_counts.sum():,}")
```

### 06. Trong bộ dữ liệu có bao nhiêu tên khác nhau?

```python
# Đếm số lượng tên khác nhau trong bộ dữ liệu
unique_names = df['Name'].nunique()
print(f"Số lượng tên khác nhau trong bộ dữ liệu: {unique_names:,}")
```

### 07. Tên nào xuất hiện ít nhất?

```python
# Tìm tên xuất hiện ít nhất
name_counts = df.groupby('Name')['Count'].sum().sort_values()
min_count = name_counts.min()
least_common_names = name_counts[name_counts == min_count]

print(f"Số lần xuất hiện ít nhất: {min_count}")
print(f"Số lượng tên có tần suất thấp nhất: {len(least_common_names)}")
print("\nMột số tên xuất hiện ít nhất:")
print(least_common_names.head(10))
```

### 08. Top 10 tên được sử dụng nhiều nhất?

```python
# Top 10 tên được sử dụng nhiều nhất
top_10_names = df.groupby('Name')['Count'].sum().sort_values(ascending=False).head(10)
print("Top 10 tên được sử dụng nhiều nhất:")
for i, (name, count) in enumerate(top_10_names.items(), 1):
    print(f"{i:2d}. {name:<12} : {count:,}")

print(top_10_names)
```

## TRỰC QUAN DỮ LIỆU

### 09. Vẽ đồ thị top 10 tên xuất hiện nhiều nhất năm 2014

```python
import matplotlib.pyplot as plt

# Lọc dữ liệu năm 2014
df_2014 = df[df['Year'] == 2014]

# Tính top 10 tên xuất hiện nhiều nhất năm 2014
top_10_2014 = df_2014.groupby('Name')['Count'].sum().sort_values(ascending=False).head(10)

# Vẽ đồ thị
plt.figure(figsize=(12, 8))
bars = plt.bar(range(len(top_10_2014)), top_10_2014.values, color='skyblue')
plt.xlabel('Tên')
plt.ylabel('Số lượng')
plt.title('Top 10 tên xuất hiện nhiều nhất năm 2014')
plt.xticks(range(len(top_10_2014)), top_10_2014.index, rotation=45)

# Thêm giá trị trên mỗi cột
for i, bar in enumerate(bars):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 500, 
             f'{top_10_2014.values[i]:,}', ha='center', va='bottom')

plt.tight_layout()
plt.show()

# In kết quả
print("Top 10 tên xuất hiện nhiều nhất năm 2014:")
for i, (name, count) in enumerate(top_10_2014.items(), 1):
    print(f"{i:2d}. {name:<12} : {count:,}")
```

### 10. Vẽ đồ thị thể hiện tỷ lệ nam nữ qua các năm 2004-2014

```python
# Tính tỷ lệ phần trăm nam nữ qua các năm
gender_ratio_by_year = df.groupby(['Year', 'Gender'])['Count'].sum().unstack()
gender_ratio_percent = gender_ratio_by_year.div(gender_ratio_by_year.sum(axis=1), axis=0) * 100

# Vẽ đồ thị
plt.figure(figsize=(12, 6))
plt.plot(gender_ratio_percent.index, gender_ratio_percent['F'], marker='o', label='Nữ', linewidth=2)
plt.plot(gender_ratio_percent.index, gender_ratio_percent['M'], marker='s', label='Nam', linewidth=2)

plt.xlabel('Năm')
plt.ylabel('Tỷ lệ (%)')
plt.title('Tỷ lệ nam nữ qua các năm 2004-2014')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xticks(range(2004, 2015))
plt.ylim(40, 60)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Tạo dữ liệu mẫu để minh họa
sample_data = {
    'Year': [2004, 2004, 2005, 2005, 2006, 2006],
    'Gender': ['F', 'M', 'F', 'M', 'F', 'M'],
    'Count': [1500000, 1600000, 1520000, 1580000, 1540000, 1590000]
}
df_sample = pd.DataFrame(sample_data)

print("=== DỮ LIỆU GỐC ===")
print(df_sample)
print()

# BƯỚC 1: df.groupby(['Year', 'Gender'])['Count'].sum()
print("=== BƯỚC 1: Nhóm theo Year và Gender, tính tổng ===")
grouped = df_sample.groupby(['Year', 'Gender'])['Count'].sum()
print("Kết quả groupby (MultiIndex Series):")
print(grouped)
print()

# BƯỚC 2: .unstack()
print("=== BƯỚC 2: Chuyển từ MultiIndex Series thành DataFrame ===")
gender_ratio_by_year = grouped.unstack()
print("Sau khi unstack():")
print(gender_ratio_by_year)
print()

# BƯỚC 3: .sum(axis=1) - Tính tổng mỗi hàng
print("=== BƯỚC 3: Tính tổng mỗi năm (F + M) ===")
row_totals = gender_ratio_by_year.sum(axis=1)
print("Tổng mỗi năm:")
print(row_totals)
print()

# BƯỚC 4: .div(..., axis=0) - Chia để tính tỷ lệ
print("=== BƯỚC 4: Chia mỗi giá trị cho tổng của năm đó ===")
ratios = gender_ratio_by_year.div(row_totals, axis=0)
print("Tỷ lệ thập phân:")
print(ratios.round(4))
print()

# BƯỚC 5: * 100 - Chuyển sang phần trăm
print("=== BƯỚC 5: Chuyển sang phần trăm ===")
gender_ratio_percent = ratios * 100
print("Tỷ lệ phần trăm:")
print(gender_ratio_percent.round(2))
print()

# HÌNH DUNG TRỰC QUAN
print("=== HÌNH DUNG TRỰC QUAN ===")
print("Năm 2004:")
print(f"  Nữ (F): {gender_ratio_by_year.loc[2004, 'F']:,} / {row_totals[2004]:,} = {gender_ratio_percent.loc[2004, 'F']:.1f}%")
print(f"  Nam (M): {gender_ratio_by_year.loc[2004, 'M']:,} / {row_totals[2004]:,} = {gender_ratio_percent.loc[2004, 'M']:.1f}%")
print()

print("Năm 2005:")
print(f"  Nữ (F): {gender_ratio_by_year.loc[2005, 'F']:,} / {row_totals[2005]:,} = {gender_ratio_percent.loc[2005, 'F']:.1f}%")
print(f"  Nam (M): {gender_ratio_by_year.loc[2005, 'M']:,} / {row_totals[2005]:,} = {gender_ratio_percent.loc[2005, 'M']:.1f}%")
```
