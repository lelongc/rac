### Bài 1

Nghiên cứu mức thu nhập X($)  và  chi tiêu Y($) trong một tháng của một công ty. Khảo sát ngẫu nhiên 7 nhân viên, ta thu được bảng số liệu sau:

| X            | 130     | 150      | 133      | 170      | 170      | 210      | 230      |
| :----------: | :-----: | :------: | :------: | :------: | :------: | :------: | :------: |
| **Y**        | **84**  | **120**  | **108**  | **130**  | **130**  | **150**  | **160**  |


a. Dựa vào bảng dữ liệu trên cho biết có xây dựng được mô hình hồi quy hay không? Nếu có hãy xây dựng mô hình hồi quy để ước tính chi tiêu dựa theo thu nhập

b. Dự đoán mức chi tiêu của nhân viên công ty trên nếu biết thu nhập là 200$

```python
# Bài 1: Hồi quy tuyến tính Y ~ X và dự đoán tại X = 200
import numpy as np
import statsmodels.api as sm
from scipy import stats

X = np.array([130, 150, 133, 170, 170, 210, 230], dtype=float)
Y = np.array([ 84, 120, 108, 130, 130, 150, 160], dtype=float)

# Tương quan
r, p_corr = stats.pearsonr(X, Y)

# Hồi quy tuyến tính với hệ số chặn
X1 = sm.add_constant(X)
model = sm.OLS(Y, X1).fit()

print("Bài 1 - Hồi quy tuyến tính Y ~ X")
print(f"Pearson r = {r:.4f}, p-value = {p_corr:.4g}")
print(f"Intercept = {model.params[0]:.4f}, Slope = {model.params[1]:.4f}")
print(f"R^2 = {model.rsquared:.4f}, F = {model.fvalue:.4f}, p(F) = {model.f_pvalue:.4g}")
print("Kết luận (5%): " + ("Có thể xây dựng mô hình (slope có ý nghĩa)." if model.pvalues[1] < 0.05 else "Không đủ bằng chứng về mối liên hệ tuyến tính."))

# Dự đoán tại X = 200
x_new = np.array([[1.0, 200.0]])
pred_sf = model.get_prediction(x_new).summary_frame(alpha=0.05)
y_hat = float(pred_sf["mean"][0])
ci_l, ci_u = float(pred_sf["mean_ci_lower"][0]), float(pred_sf["mean_ci_upper"][0])
pi_l, pi_u = float(pred_sf["obs_ci_lower"][0]), float(pred_sf["obs_ci_upper"][0])
print(f"Dự đoán Y khi X=200: {y_hat:.4f}")
print(f"95% CI (mean): [{ci_l:.4f}, {ci_u:.4f}]")
print(f"95% PI (obs):  [{pi_l:.4f}, {pi_u:.4f}]")
```

### Bài 2

Tiến hành nghiên cứu mối quan hệ giữa:

- X: giá bán kẻ của thịt gà (nghìn/kg) 
- Y: lượng thịt gà tiêu thụ (kg/tháng)

Dữ liệu thu được gồm 10 tháng như sau:

| X            | 35       | 40       | 45       | 40       | 38       | 40       | 45       | 30       | 42       | 38       |
| :----------: | :------: | :------: | :------: | :------: | :------: | :------: | :------: | :------: | :------: | :------: |
| **Y**        | **2.29** | **2.04** | **0.59** | **2.06** | **2.50** | **2.32** | **1.57** | **3.96** | **1.43** | **2.34** |


a. Dựa vào bảng dữ liệu trên cho biết có xây dựng được mô hình hồi quy hay không? Nếu có hãy xây dựng mô hình hồi quy để ước tính lượng tiêu thụ thịt gà theo giá bán.

b. Dự đoán lượng tiêu thụ khi giá bán lẻ là 41(kg/tháng)

```python
# Bài 2: Hồi quy tuyến tính Y ~ X và dự đoán tại X = 41
import numpy as np
import statsmodels.api as sm
from scipy import stats

X = np.array([35, 40, 45, 40, 38, 40, 45, 30, 42, 38], dtype=float)
Y = np.array([2.29, 2.04, 0.59, 2.06, 2.50, 2.32, 1.57, 3.96, 1.43, 2.34], dtype=float)

# Tương quan
r, p_corr = stats.pearsonr(X, Y)

# Hồi quy tuyến tính với hệ số chặn
X1 = sm.add_constant(X)
model = sm.OLS(Y, X1).fit()

print("Bài 2 - Hồi quy tuyến tính Y ~ X")
print(f"Pearson r = {r:.4f}, p-value = {p_corr:.4g}")
print(f"Intercept = {model.params[0]:.4f}, Slope = {model.params[1]:.4f}")
print(f"R^2 = {model.rsquared:.4f}, F = {model.fvalue:.4f}, p(F) = {model.f_pvalue:.4g}")
print("Kết luận (5%): " + ("Có thể xây dựng mô hình (slope có ý nghĩa)." if model.pvalues[1] < 0.05 else "Không đủ bằng chứng về mối liên hệ tuyến tính."))

# Dự đoán tại X = 41
x_new = np.array([[1.0, 41.0]])
pred_sf = model.get_prediction(x_new).summary_frame(alpha=0.05)
y_hat = float(pred_sf["mean"][0])
ci_l, ci_u = float(pred_sf["mean_ci_lower"][0]), float(pred_sf["mean_ci_upper"][0])
pi_l, pi_u = float(pred_sf["obs_ci_lower"][0]), float(pred_sf["obs_ci_upper"][0])
print(f"Dự đoán Y khi X=41: {y_hat:.4f}")
print(f"95% CI (mean): [{ci_l:.4f}, {ci_u:.4f}]")
print(f"95% PI (obs):  [{pi_l:.4f}, {pi_u:.4f}]")
```
