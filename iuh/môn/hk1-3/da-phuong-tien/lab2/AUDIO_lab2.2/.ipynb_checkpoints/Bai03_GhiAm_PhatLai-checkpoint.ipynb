# Bài 3: Ghi âm từ micro và phát lại (`sounddevice`)

**Mục tiêu:**
- Ghi âm từ micro trong vài giây.
- Lưu thành `.wav`.
- Phát lại âm thanh đã ghi.

**Lưu ý:** Cần có thiết bị âm thanh.
**Thư viện:** `sounddevice`, `scipy.io.wavfile`.

```python
#@title Cài đặt (nếu cần)
# !pip install sounddevice scipy
```

```python
import sounddevice as sd
# sounddevice: Thư viện cho phép ghi âm (microphone) và phát lại âm thanh qua loa, dễ dùng cho xử lý âm thanh thời gian thực.

import numpy as np
# numpy: Thư viện xử lý số, tạo mảng, chuyển đổi kiểu dữ liệu, rất nhanh và tiện lợi.

from scipy.io.wavfile import write, read
# scipy.io.wavfile: Thư viện giúp đọc và ghi file WAV dễ dàng (chuẩn âm thanh số phổ biến).

fs = 44100  # Tần số lấy mẫu (Hz). 44100Hz là chuẩn CD, nghĩa là mỗi giây lấy 44100 mẫu âm thanh.
seconds = 3  # Thời lượng ghi âm (tính bằng giây). Ở đây sẽ ghi âm 3 giây.

print("Đang ghi âm...")  # Thông báo cho người dùng biết quá trình bắt đầu ghi âm

audio = sd.rec(
    int(seconds * fs),    # Tổng số mẫu cần ghi (thời gian * tần số lấy mẫu)
    samplerate=fs,        # Đặt tần số lấy mẫu
    channels=1,           # Ghi âm 1 kênh (mono). Nếu muốn stereo, bạn đặt là 2.
    dtype='int16'         # Kiểu dữ liệu số nguyên 16-bit (chuẩn cho WAV 16 bit)
)
# sd.rec(...): Bắt đầu ghi âm từ micro, trả về mảng numpy chứa âm thanh.

sd.wait()  # Chờ cho đến khi ghi âm xong (đảm bảo mọi dữ liệu đã thu được lưu lại)
print("Hoàn tất ghi âm!")  # Thông báo đã xong quá trình ghi âm

write("my_record.wav", fs, audio)
# write(...): Ghi mảng dữ liệu audio ra file WAV với tần số lấy mẫu fs.
# File my_record.wav sẽ lưu lại đoạn âm thanh vừa ghi.

print("Đã lưu: my_record.wav")  # Thông báo đường dẫn file vừa lưu
```

### Phát lại file vừa ghi

```python
# Phát lại (nếu môi trường cho phép phát âm)
rate, data = read("my_record.wav")
sd.play(data, rate)
sd.wait()
print("Phát lại xong!")
```
