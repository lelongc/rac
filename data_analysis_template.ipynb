- Câu 6: `VIP` → thay bằng giá trị đặc biệt trong cột của bạn
- Câu 7: `Khu vực` → thay bằng tên cột và mô tả của bạn
- Câu 8: `Sản phẩm` → thay bằng đối tượng cần thống kê của bạn
- Câu 9: `Nhóm tuổi` → thay bằng cột nhóm của bạn
- Câu 10: `đơn hàng` và `giá trị` → thay bằng đối tượng và metric của bạn

````python
import pandas as pd
import numpy as np

# 1. Đọc dữ liệu - THAY: 'sales_data.csv' → tên file của bạn
df = pd.read_csv('data/sales_data.csv')

# 2. Xóa cột không cần thiết
if 'Unnamed: 0' in df.columns:
    df = df.drop('Unnamed: 0', axis=1)

print("Dữ liệu sau khi đọc:")
print(df.head())
print("\n" + "="*50)

# 3. Cho biết dữ liệu có bao nhiêu dòng, bao nhiêu thuộc tính
print(f"3. Dữ liệu có {df.shape[0]} dòng và {df.shape[1]} thuộc tính")
print("\n" + "="*50)

# 4. Phân loại thuộc tính
print("4. Phân loại thuộc tính:")
quantitative_cols = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
print("Thuộc tính định lượng:", quantitative_cols)
print("Thuộc tính phân loại:", categorical_cols)
print("\n" + "="*50)

# 5. Thống kê theo cột phân loại 1 - THAY: 'gender' và 'Male'/'Female'
print("5. Thống kê theo giới tính:")
gender_counts = df['gender'].value_counts()
gender_ratio = df['gender'].value_counts(normalize=True) * 100
print(f"Nam: {gender_counts.get('Male', 0)} người ({gender_ratio.get('Male', 0):.1f}%)")
print(f"Nữ: {gender_counts.get('Female', 0)} người ({gender_ratio.get('Female', 0):.1f}%)")
print("\n" + "="*50)

# 6. Thống kê theo cột phân loại 2 - THAY: 'customer_type' và 'VIP'/'Regular'
print("6. Thống kê loại khách hàng:")
type_counts = df['customer_type'].value_counts()
type_ratio = df['customer_type'].value_counts(normalize=True) * 100
print(f"VIP: {type_counts.get('VIP', 0)} người ({type_ratio.get('VIP', 0):.1f}%)")
print(f"Thường: {type_counts.get('Regular', 0)} người ({type_ratio.get('Regular', 0):.1f}%)")
print("\n" + "="*50)

# 7. Thống kê theo cột có nhiều giá trị - THAY: 'region'
print("7. Thống kê theo khu vực:")
region_counts = df['region'].value_counts()
region_ratio = df['region'].value_counts(normalize=True) * 100
print("Thứ tự từ nhiều đến ít:")
for region in region_counts.index:
    print(f"{region}: {region_counts[region]} lượt ({region_ratio[region]:.1f}%)")
print(f"Khu vực có khách nhiều nhất: {region_counts.index[0]}")
print("\n" + "="*50)

# 8. Thống kê sản phẩm/dịch vụ - THAY: 'product_name'
print("8. Thống kê sản phẩm bán chạy:")
product_counts = df['product_name'].value_counts()
product_ratio = df['product_name'].value_counts(normalize=True) * 100
print("Top 5 sản phẩm bán chạy:")
for i, product in enumerate(product_counts.head().index):
    print(f"{i+1}. {product}: {product_counts[product]} lượt ({product_ratio[product]:.1f}%)")
print(f"Sản phẩm bán chạy nhất: {product_counts.index[0]}")
print("\n" + "="*50)

# 9. Thống kê nhóm số - THAY: 'age_group'
print("9. Thống kê theo nhóm tuổi:")
age_counts = df['age_group'].value_counts().sort_index()
age_ratio = df['age_group'].value_counts(normalize=True).sort_index() * 100
print("Thống kê theo nhóm:")
for age in age_counts.index:
    print(f"{age}: {age_counts[age]} người ({age_ratio[age]:.1f}%)")
most_common_age = df['age_group'].value_counts().index[0]
print(f"Nhóm tuổi đông nhất: {most_common_age}")
print("\n" + "="*50)

# 10. Trung bình giá trị - THAY: 'sales_amount'
print("10. Trung bình giá trị đơn hàng:")
avg_sales = df['sales_amount'].mean()
print(f"Trung bình: {avg_sales:.2f} VND")
````

## YẾU TỐ ẢNH HƯỞNG ĐẾN DOANH SỐ

Trong phần này, chúng ta thử tìm hiểu về doanh số bán hàng và các yếu tố có thể ảnh hưởng đến nó

**HƯỚNG DẪN THAY ĐỔI:**
- `DOANH SỐ` → thay bằng biến mục tiêu của bạn (tip, lương, điểm số, v.v.)

1. Tỷ lệ có doanh số của khách hàng là bao nhiêu?
2. Tỷ lệ doanh số trên tổng chi tiêu là bao nhiêu?
3. Giữa Nam và Nữ ai có doanh số nhiều hơn, ai mua hàng thường xuyên hơn?
4. Giữa khách VIP và khách thường ai có doanh số nhiều hơn, ai mua hàng thường xuyên hơn?
5. Thống kê tổng doanh số theo tháng.

**HƯỚNG DẪN THAY ĐỔI:**
- Câu 1: `có doanh số` → thay điều kiện phù hợp (>0, không null, v.v.)
- Câu 2: `doanh số/chi tiêu` → thay 2 cột tương ứng của bạn
- Câu 3-4: Thay nhóm so sánh của bạn
- Câu 5: `tháng` → thay cột thời gian của bạn

````python
# YẾU TỐ ẢNH HƯỞNG ĐẾN DOANH SỐ

print("YẾU TỐ ẢNH HƯỞNG ĐẾN DOANH SỐ")
print("="*50)

# 1. Tỷ lệ có doanh số - THAY: 'sales_amount' và điều kiện > 0
print("1. Tỷ lệ có doanh số:")
has_sales = (df['sales_amount'] > 0).sum()
total_records = len(df)
sales_rate = (has_sales / total_records) * 100
print(f"Số có doanh số: {has_sales}/{total_records}")
print(f"Tỷ lệ có doanh số: {sales_rate:.1f}%")
print("\n" + "="*50)

# 2. Tỷ lệ doanh số trên tổng chi tiêu - THAY: 'sales_amount' và 'total_spending'
print("2. Tỷ lệ doanh số trên tổng chi tiêu:")
df['sales_percentage'] = (df['sales_amount'] / df['total_spending']) * 100
avg_sales_percentage = df['sales_percentage'].mean()
median_sales_percentage = df['sales_percentage'].median()
print(f"Tỷ lệ doanh số trung bình: {avg_sales_percentage:.2f}%")
print(f"Tỷ lệ doanh số trung vị: {median_sales_percentage:.2f}%")
print("\n" + "="*50)

# 3. So sánh doanh số giữa Nam và Nữ - THAY: 'gender'
print("3. So sánh doanh số giữa Nam và Nữ:")
sales_by_gender = df.groupby('gender').agg({
    'sales_amount': ['mean', 'sum', 'count'],
    'sales_percentage': 'mean'
}).round(2)
print("Bảng thống kê sales_by_gender:")
print(sales_by_gender)

print("Doanh số trung bình:")
for gender in sales_by_gender.index:
    print(f"{gender}: {sales_by_gender.loc[gender, ('sales_amount', 'mean')]:,.2f} VND")

print("\nTổng doanh số:")
for gender in sales_by_gender.index:
    print(f"{gender}: {sales_by_gender.loc[gender, ('sales_amount', 'sum')]:,.2f} VND")

print("\nTỷ lệ doanh số trung bình:")
for gender in sales_by_gender.index:
    print(f"{gender}: {sales_by_gender.loc[gender, ('sales_percentage', 'mean')]:.2f}%")
print("\n" + "="*50)

# 4. So sánh doanh số giữa VIP và thường - THAY: 'customer_type'
print("4. So sánh doanh số giữa khách VIP và thường:")
sales_by_type = df.groupby('customer_type').agg({
    'sales_amount': ['mean', 'sum', 'count'],
    'sales_percentage': 'mean'
}).round(2)
print("Bảng thống kê sales_by_type:")
print(sales_by_type)

print("Doanh số trung bình:")
for ctype in sales_by_type.index:
    print(f"{ctype}: {sales_by_type.loc[ctype, ('sales_amount', 'mean')]:,.2f} VND")

print("\nTổng doanh số:")
for ctype in sales_by_type.index:
    print(f"{ctype}: {sales_by_type.loc[ctype, ('sales_amount', 'sum')]:,.2f} VND")

print("\nTỷ lệ doanh số trung bình:")
for ctype in sales_by_type.index:
    print(f"{ctype}: {sales_by_type.loc[ctype, ('sales_percentage', 'mean')]:.2f}%")
print("\n" + "="*50)

# 5. Thống kê tổng doanh số theo tháng - THAY: 'month' và thứ tự tháng
print("5. Tổng doanh số theo tháng:")
sales_by_month = df.groupby('month').agg({
    'sales_amount': ['sum', 'mean', 'count']
}).round(2)
print("Bảng thống kê sales_by_month:")
print(sales_by_month)

# Sắp xếp theo thứ tự tháng - THAY thứ tự phù hợp với dữ liệu
month_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales_by_month = sales_by_month.reindex([month for month in month_order if month in sales_by_month.index])
print(sales_by_month)
print("\n" + "="*50)

print("Thống kê chi tiết:")
for month in sales_by_month.index:
    total_sales = sales_by_month.loc[month, ('sales_amount', 'sum')]
    avg_sales = sales_by_month.loc[month, ('sales_amount', 'mean')]
    count = sales_by_month.loc[month, ('sales_amount', 'count')]
    print(f"{month}: Tổng {total_sales:,.2f} | TB {avg_sales:,.2f} | {count} đơn")

print(f"\nTháng có tổng doanh số cao nhất: {sales_by_month[('sales_amount', 'sum')].idxmax()} ({sales_by_month[('sales_amount', 'sum')].max():,.2f})")
print(f"Tháng có doanh số trung bình cao nhất: {sales_by_month[('sales_amount', 'mean')].idxmax()} ({sales_by_month[('sales_amount', 'mean')].max():,.2f})")
````
